```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Days': ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'],
    'Study_Hours': [2,3,4,6,5],
    'Marks' : [50,60,55,70,65],
    'Attendance': [60,70,65,80,75],
    'Sleep_Hours': [6,7,6,8,7],
}
df = pd.DataFrame(data)
df
```

```python
#1. Line Chart (Trend)
df = pd.DataFrame(data)
plt.plot(df['Days'], df['Study_Hours'],marker='^')
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.show()
```

```python
# Advanced Line Chart
plt.figure(figsize=(7,4))

plt.plot(df['Days'], df['Study_Hours'], marker='*',color='green', label='Study Hours')

plt.plot(df['Days'], df['Marks'], marker='o', color='blue', label='Marks')

plt.title('Study Hours and Marks trend')
plt.xlabel('Days')
plt.ylabel('Values')
plt.legend()
```

```python
#2. Bar Chart (Comparison)

plt.bar(df['Days'], df['Marks'], color='Orange')

plt.xlabel('Days')
plt.ylabel('Marks')
plt.show()
```

```python
plt.bar(df['Study_Hours'], df['Marks'], color='Green')

plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
#Advanced Bar Graph
bars = plt.bar(df['Days'], df['Marks'], color='Violet')

for bar in bars:
  y = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2, y, str(y),ha='center', va='bottom')
plt.title('Marks Comparison')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.grid(axis='both')
plt.show()
```

```python
#3.Histogram(Dsitribution)
#Shows how values are spread

data = {
    'Marks' : [50,60,65,80,90,80,95,85,75,60,55,85]
}
df = pd.DataFrame(data)
plt.hist(df['Marks'], bins=5, edgecolor='Red',alpha=0.5)
plt.title('Marks Distribution')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
plt
```

```python
#4. Scatter Plot (Relationship)
import matplotlib.pyplot as plt


plt.scatter(df['Study_Hours'], df['Marks'])

plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
# Conditional scatter plot
# Example category

df['Result'] = ['Fail','Pass','Fail','Pass','Pass']
colors = ['red' if r=='Fail' else 'green' for r in df['Result']]

plt.scatter(df['Study_Hours'], df['Marks'], c = colors)

plt.title('Study Hours vs Marks(Result-wise)')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```
