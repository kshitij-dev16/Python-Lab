```python
import pandas as pd
import numpy as np

data = {
    'Product':['Laptop','Mobile','Tablet','Heaphopnes','Smartwatch','Camera'],
    'Price' : [55000,20000,30000,2000,8000,45000],
    'Units_sold' : [25,60,35,80,50,15],
    'Discount' : [10,15,12,20,18,8]
}

df = pd.DataFrame(data)

#1. Min-Max Normalization
# Min-Max Normalization for Price

df['Price_MinMax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max()
  -df['Price'].min())


print("\nMin-Max Normalized Price:")
print(df[['Product','Price',"Price_MinMax"]])

#2. Min-Max Normalization
# Min-Max Normalization for Discount

df['Price_MinMax'] = (df['Discount'] - df['Discount'].min()) / (df['Discount'].max()
  -df['Price'].min())


print("\nMin-Max Normalized Price:")
print(df[['Product','Discount',"Price_MinMax"]])



```

```python
# Normalization Multiple Columns

cols = ['Price','Units_sold','Discount']

df_norm = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())

print("\nNormalized Dataset")
df_norm
```

```python
#2.  Z-score Normalization
df['Untis_Zscore'] = (df['Units_sold'] - df['Units_sold'].mean()) / df['Units_sold'].std()

print("\nZ-score Normalized Units_sold:")
print(df[['Product','Units_sold','Untis_Zscore']])
```

```python
#3. Decimal Scaling
df['Price_Decimal'] = df['Price'] / 10000

print("\nDecimal Scaled Price:")
print(df[['Product','Price','Price_Decimal']])

```

Part-B Turning categorical variables into quantitative variables in Python

```python
import pandas as pd
import numpy as np

data = {
    'Order_ID' : [101,102,103,104,105,106],
    'Customer_Gender' : ['Male','Female','Male','Female','Male','Female'],
    'Payment_Method' : ['UPI','Credit Card','Debit card','COD','UPI','Debit Card'],
    'Product_Category' : ['Electronics','Clothing','Electronics','Clothing','Electronics','Clothing'],
    'City' : ['Mumbai','Delhi','Mumbai','Delhi','Mumbai','Delhi'],
    'Order_Value' : [2500,1500,3000,1000,4100,5000]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Method 1 : Label Encoding
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Gender_Label'] = le.fit_transform(df['Customer_Gender'])

print(df)
```

```python
#Method 2 : One-Hot Encoding

df_encoded = pd.get_dummies(df,columns=['Payment_Method'])

print(df_encoded)
```

```python
#Method 3: Dummy Encoding (Drop First)

df_dummy = pd.get_dummies(df,columns=['Product_Category'],drop_first=True)

print(df_dummy)
```
