```python
# while loop
# Used when the number of iterations is not known
Syntax
while condition:
  statement(s)
```

```python
# Print numbers from 1 to 5
i = 1
while(i<=5):
  print(i)
  i = i + 1
```

```python
# Take input from the user to print the number of iterations

i = 1
num = int(input("Enter the number of iterations: "))
while (i<=num):
  print(i, end = " ")
  i = i+1
```

```python
# To calculate the factorial of a number

n = int(input("Enter a number: "))
fact = 1
i = 1
while (i <= n):
  fact = fact * i
  i = i + 1
print("The factorial of",n,"is",fact)
```

```python
# Fibonacci series using while loop

n = int(input("Enter the number of terms: "))

a = 1
b = 2
i = 0

while i < n:
    print(a, end=" ")
    a, b = b, a + b
    i += 1
```

```python
# Reverse a number

limit = int(input("Enter the number to reverse: "))
rev = 0
num = limit # Initialize 'num' with the input 'limit'

while num > 0:
  digit = num % 10
  rev = rev * 10 + digit
  num = num // 10
print("Reversed number =", rev)
```

```python
# Check Palindrome number/string

st = input("Enter a string: ")
rev = st[::-1]
if st == rev:
  print("Palindrome")
else:
  print("Not Palindrome")


```

```python
# Count digits in a number

num = int(input("Enter a number: "))
count = 0
while num > 0:
  count += 1
  num//=10

print("Number of digits =", count)

```

```python
# Exit the loop when i is 3

i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```
