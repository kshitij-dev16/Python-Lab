```python
import pandas as pd
import numpy as np
#Step1
df = pd.read_csv('student_data.csv')
print("Original Data")
print(df)
```

```python
# Step 2 :Replac'-'with NaN to mark missing values

df.replace("-", np.nan, inplace=True)
#convert columns to numeric where needed
df["Age"]=pd.to_numeric(df["Age"],errors='coerce')
df["Marks"]=pd.to_numeric(df["Marks"],errors='coerce')

'''When converting data (like strings→ datetime),
if some values are invaid or cannot be converted,
instead of throwing an error,it will :
Replace those invalid values with NaT (not a time)'''
df
```

```python
# Step 3 : Check missing values
print ("\nMissing values")
print(df.isnull().sum())
```

```python
# fill missing age with mean
mean_age = df['Age'].mean()
df['Age'].fillna(mean_age, inplace=True)
df
```

```python
# fill missing marks with median
median_marks = df['Marks'].median()
df['Marks'].fillna(median_marks, inplace=True)
df
```

```python
print(df.isnull().sum())
```

```python
#Step 5 : Fix inconsistent text formatting
df["Department"] = df["Department"].str.upper()
df
```

```python
#Step 6: Convert Admission_Date tostandard datetime format
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"], errors='coerce')
df

```

```python
#Step 7 : show cleaned data
print("\nCleaned Data")
print(df)
```

```python
# Step 8 : Save cleared dataset
df.to_csv("cleaed_student_data.csv",index=False)

print('\nCleaned CSV fie saved as cleaned_student_data.csv')
```

```python
df = pd.read_csv('Cars93.csv')
df
```

```python
#Imputing the Missing Value
#1. Luggage Room
df ['Luggage.room'] = df['Luggage.room'].fillna(value=df['Luggage.room'].mean())
```

```python
#2.Rear.seat.room
df ['Rear.seat.room'] = df ['Rear.seat.room'].fillna(value=df['Rear.seat.room'].mean())
```

```python
#3.AirBags
df['AirBags'] = df['AirBags'].fillna(value=df['AirBags'].mode().values[0])
```

```python
#Save cleaned dataset
df.to_csv("cleaned_Cars93.csv", index=False)
```
