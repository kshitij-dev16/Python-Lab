```python
import pandas as pd

data = {
    'Order_ID':['01','02','03','04','05','06','07','08','09','010'],

    'Category':['Electronics','Clothing','Electronics','Grocery','Clothing',
                'Electronics','Grocery','Clothing','Electronics','Grocery'],

    'Payment_Method':['UPI','Card','UPI','Cash','Card',
                       'UPI','Cash','Card','UPI','Cash'],

    'Delivery_Type':['Express','Standard','Express','Standard','Express',
                      'Standard','Express','Standard','Express','Standard'],

    'Customer_Type':['New',"Returning","Returning","New","Returning",
                      'New','Returning','New','Returning', 'New'] # Added 'New' to match length
}

df = pd.DataFrame(data)

print(df)
```

```python
#1. Frequency Count
#Count how many orders in each category.


print(df['Category'].value_counts())
print()
print(df['Payment_Method'].value_counts())
print()
print(df['Delivery_Type'].value_counts())
print()
print(df['Customer_Type'].value_counts())
print()

```

```python
#2. Unique Values
#Display all unique categories

print(df['Category'].unique())
print()
print(df['Payment_Method'].unique())
print()
print(df['Delivery_Type'].unique())
print()
print(df['Customer_Type'].unique())
print()
```

```python
#3. Number of unique Values
print(df['Category'].nunique())
print()
print(df['Payment_Method'].nunique())
print()
print(df['Delivery_Type'].nunique())
print()
print(df['Customer_Type'].nunique())
```

```python
#4. Cross Tabulation
#Category vs Payment Method

print(pd.crosstab(df['Category'],df['Payment_Method']))
print()
print(pd.crosstab(df['Category'],df['Delivery_Type']))
print()
print(pd.crosstab(df['Category'],df['Customer_Type']))
print()
```

```python
#5. Cross Tabulation (Delivery vs Customer Type)
print(pd.crosstab(df['Delivery_Type'],df['Customer_Type']))
print()
print(pd.crosstab(df['Payment_Method'],df['Customer_Type']))
print()
print(pd.crosstab(df['Payment_Method'],df['Delivery_Type']))
print()


```

```python
#6. Percentage Distribution
print(df['Category'].value_counts(normalize=True)*100)
print()
print(df['Payment_Method'].value_counts(normalize=True)*100)
print()
print(df['Delivery_Type'].value_counts(normalize=True)*100)
print()
print(df['Customer_Type'].value_counts(normalize=True)*100)
print()
```

```python
#8. Filtering Data
#Example: Orders from Electronics category

electronics_orders = df[df['Category'] == 'Electronics']
print(electronics_orders)
print()

```

```python
#9. Grouping
#Orders by categroy and payment method

print(df.groupby(['Category','Payment_Method']).value_counts())
print()
print(df.groupby(['Category','Delivery_Type']).value_counts())
print()
print(df.groupby(['Category','Customer_Type']).value_counts())
print()
```

```python
#10. Sorting Data
sorted_df = df.sort_values(by='Category')
print(sorted_df)
print()
```
